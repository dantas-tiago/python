Uma variável é como uma etiqueta (nome) que usamos para guardar informações em uma posição de memória.

- Pelo PEP8 (https://peps.python.org/pep-0008/) nomes de variáveis devem ser minúsculo e, quando for nome composto, deve ser utilizado o underscore. Ex.: nome_variavel (snake case)
- Utilizar nomes sugestivos (minha_carteira)
- Quando estiver salvando uma constante, deve ser em letras MAIÚSCULAS e, quando for nome composto, deve ser utilizado o underscore. Ex.: NOME_CONSTANTE
- Evitar nome de palavras reservadas: and, class, elif, finally, if, lambda, or, TRUE, yield, as, continue, else, for, import, None, pass, try, assert, def, except, from, in, nonlocal, raise, while, break, del, FALSE, global, is, not, return, with.
- Variáveis são, no python, de tipagem dinâmica e forte:
- Tipagem Dinâmica: O tipo de dado de uma variável é determinado automaticamente pelo valor que ela armazena, e pode mudar durante a execução do programa;
- Tipagem Forte: Mesmo com tipagem dinâmica, Python não permite operações entre tipos incompatíveis sem conversão explícita.

```python
# variável numérica int
var_num_int = 1
print(var_num_int)
print(type(var_num_int))
```

```python
# a Função id irá dizer em qual local da memória foi salvo esta variável.  id é uma função nativa
print(var_num_int)
print(id(var_num_int))
print(hex(id(var_num_int))) # função hex transforma o valor em hexadecimal
```

```python
# variável numérica float
var_num_float = 2.5
print(var_num_float)
print(id(var_num_float))
```

```python
var_nome = "Esta é uma variável string!"
print(var_nome)
```

```python
var_nome = "Esta é uma variável string!"
print(var_nome)
print(id(var_nome))
print(var_nome,"A posição de memória dessa variável é:", hex(id(var_nome))) 
# é possível imprimir mais de um texto no print, deve-se apenas colocar separado por ,
```

```python
# declaração de mais de uma variável em uma linha

nome_pessoa, idade_pessoa = "Tiago", 38
# nome_pessoa, idade_pessoa = "Tiago", 38 #opcional

print(nome_pessoa, end="\n") # por padrão ele considera \n, ou seja, uma quebra de linha
print(idade_pessoa)
```

```python
# declaração de mais de uma variável em uma linha

nome_pessoa, idade_pessoa = "Tiago", 38
# nome_pessoa, idade_pessoa = "Tiago", 38 #opcional

print(nome_pessoa, end=" - ") # caso queira colocar um outro caracter no final, vc pode usar o parâmetro end=""
print(idade_pessoa)
```

```python
# mostrando tipagem dinâmica

x = 10      # 'x' é um inteiro
print(x)
print(type(x))
x = "Olá"   # Agora, 'x' é uma string
print(x)
print(type(x))
```

```python
# mostrando tipagem forte
x = 10
print(x)
print(type(x))
y = "20"
print(y)
print(type(y))
z = x + y  # Isso gera um erro de tipo, pois você não pode somar um inteiro e uma string diretamente.
```
